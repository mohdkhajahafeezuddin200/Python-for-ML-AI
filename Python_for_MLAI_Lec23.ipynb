```python
students = [
    {
        "roll no":101,
        "name":"haa",
        "age":20,
        "marks":98,
        "course":"python"
    },
    {
        "roll no":102,
        "name":"hhfaa",
        "age":20,
        "marks":58,
        "course":"python"
    },
    {
        "roll no":103,
        "name":"h74aa",
        "age":20,
        "marks":78,
        "course":"python"
    },
    {
        "roll no":104,
        "name":"hyfaa",
        "age":20,
        "marks":78,
        "course":"python"
    },
    {
        "roll no":105,
        "name":"haasd",
        "age":21,
        "marks":38,
        "course":"python"
    }
]
```

```python
def serch_name(name):
  for s in students:
    if s["name"].lower() == name.lower():
      print(s)
      return "record found"

  return "not found"
```

```python
def toppr(students):
  max_marks = 0
  for student in students:
    if student["marks"] > max_marks:
      max_marks = student["marks"]
      topper = student["name"]
      print("topper is ",topper)
```

```python
def average(t):
  return t/len(students)
```

#Student management system

```python
print("======STUDENT MANAGEMENT SYSTEM======")

print("2. display student")
print("3. search student")
print("4. topper")
print("5. calculate avg")
print("6. exit")


option = input("enter your choice:")

if option == 6:
  print("successfully exited")

if option == 2:
  for student in students:
    print(student)

if option == 3:
  n = input("enter name")
  serch_name(n)
if option == 4:
  toppr(students)

if option ==5:
  m = int(input("enter marks"))
  average(m)
```
