```python
def count_incourse(students):
  c_pyt =0
  c_ai = 0
  c_sql =0
  for student in students:
    if student["course"] == "python":
      c_pyt += 1
    if student["course"] == "AI":
      c_ai += 1

    if student["course"] == "SQL":
      c_sql += 1

  print(c_pyt, c_ai , c_sql)
```

```python
count_incourse(students)
```
